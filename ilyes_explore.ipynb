```python
import pandas as pd

X_train = pd.read_csv("Data/train_input_Z61KlZo.csv")
y_train = pd.read_csv("Data/train_output_DzPxaPY.csv")
# X_test = pd.read_csv("Data/test_input_5qJzHrr.csv")
# y_pred_random = pd.read_csv("Data/test_output_random.csv")

print(f"Train: {X_train.shape}")
# print(f"Test: {X_test.shape}")
```

```python
import matplotlib.pyplot as plt

non_zero_charge = y_train[y_train['CHARGE'] != 0]['CHARGE']
non_zero_charge.hist(bins=100)
plt.title("Non Zero 'CHARGE'")

non_zero_charge.describe()
```

```python
import pandas as pd

# Merge X_train and y_train on 'ID'
merged = pd.merge(X_train, y_train, on='ID')

# Group by column c and calculate percentages of zero and non-zero CHARGE
def charge_zero_nonzero_percent(df, column):
    grouped = df.groupby(column)['CHARGE']
    result = grouped.apply(lambda x: pd.Series({
        'percent_zero': (x == 0).mean() * 100,
    })).reset_index()
    return result

# Replace 'c' with your actual column name
result_df = charge_zero_nonzero_percent(merged, 'EQUIPEMENT2')
result_df.sort_values('CHARGE')
```

# Mutual Info

```python
from sklearn.feature_selection import mutual_info_classif

# clean
num_cols = X_train.select_dtypes(include=['int64', 'float64']).columns.tolist().remove('ID')
cat_cols = X_train.select_dtypes(include=['object', 'category', 'bool']).columns.tolist()
X_train_clean = X_train.copy()
X_train_clean[num_cols] = X_train[num_cols].astype("float32")
X_train_clean[cat_cols] = X_train[cat_cols].astype(str)

y_train_binary = (y_train['CHARGE'] != 0).astype(int)
mi = mutual_info_classif(X_train_clean, y_train_binary, discrete_features='auto')
mi_scores = pd.Series(mi, index=X_train_clean.columns).sort_values(ascending=False)

mi_scores[mi_scores>0.05]
```
